```python
import requests
```

```python
url = 'https://crew-ai-learning-.crewai.com'
bearer_token = ''
headers = {
    'Authorization': f'Bearer {bearer_token}',
    'Content': 'application/json'
}
```

```python
response = requests.get(f'{url}/inputs', headers=headers)
if response.status_code == 200:
    print(response.json())
else:
    print(f'Error {response.status_code}')
```

```python
body = {
    'inputs':{
        'topic': 'CrewAI',
        'current_year': '2026'
    }
}

response_kickoff = requests.post(f'{url}/kickoff', headers=headers, json=body)
if response_kickoff.status_code == 200:
    print(response_kickoff.json())
else:
    print(f'Error {response_kickoff.status_code}')
```

```python
kickoff_id = '0355af6e-a24a-4dea-b0b4-3ffd3ab5bb3f'

response_status = requests.get(f'{url}/status/{kickoff_id}', headers=headers)
if response_status.status_code == 200:
    print(response_status.json())
else:
    print(f'Error {response_status.status_code}')
```

```python
raw_result = response_status.json()
result = raw_result['result']
print(result)
```
